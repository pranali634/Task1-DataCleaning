```python
import pandas as pd
```

```python
df = pd.read_csv("dataset.csv")
```

```python
print("First 5 Rows:")
print(df.head())
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()
```

```python
df = df.fillna("Unknown")
```

```python
df.columns = df.columns.str.lower().str.replace(" ", "_")
```

```python
print("\nData Types:")
print(df.dtypes)
```

```python
df.to_csv("cleaned_dataset.csv", index=False)
```
